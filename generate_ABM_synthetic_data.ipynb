```python
import ima_takeda as tkd
import numpy as np
import pandas as pd
```

# Sample: run simulations n_runs times for different combinations of E:T and record

```python
for E0 in [100, 1000, 10000]:
    for T0 in [100, 1000, 10000]:
        sim_config: dict = dict(
            E_0=E0,
            T_0=T0,
            r=0.01,
            m=0.02,
            N=96,
            L=1.0,
            k_on=0.4,
            k_off=0.3,
            k_kill=0.007,
            g_E=0,
            g_T=0.007,
            early_stop=(False, 0, 0.0),
            log_all=True,
            use_tqdm=False,
            seed=42,
        )
        print(E0, T0)
        monte_carlo, monte_carlo_baseline = tkd.single_scenario_monte_carlo(n_runs=100, sim_kwargs=sim_config)
        E = np.nanmedian(monte_carlo["E"], axis=0)
        T = np.nanmedian(monte_carlo["T"], axis=0)
        C = np.nanmedian(monte_carlo["C"], axis=0)
        pd.DataFrame({"E": E, "T": T, "C": C}).to_csv(f"data/E-{E0}_T-{T0}_AGG.csv", index=True)
```
